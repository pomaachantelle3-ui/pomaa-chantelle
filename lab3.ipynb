```python
Goal = (0, 1, 2, 3, 4, 5, 6, 7, 8)
MOVES = {"Up": -3, "Down": 3, "Left":
         -1, "Right": 1}

def neighbours(s) :
    i = s.index(0); 
    r, c = divmod(i, 3)
    
    for a, d in MOVES.items():
        if(a == "Up" and r == 0) or \
        (a == "Down" and r == 2) or \
        (a == "Left" and c == 0)or \
        (a == "Right" and c == 2):
            continue
        
        j = i +d
        t = list(s)
        t[i], t[j] =t[j], t[i]
        yield a, tuple(t)

def solvable(s):
    v = [x for x in s if x]
    inversions = sum(1 for i in range(len(v)) for j in range(i + 1, len(v)) if v [i] > v[j])
    return inversions % 2 == 0
```

```python
def h_zero(s):
    return 0
def h_misplaced(s):
    return sum(1 for k, v in enumerate(s) if v and v != Goal[k])
def h_mahattan(s):
    return sum(abs(k // 3 - v // 3) + abs(k % 3 - v % 3) for k, v in enumerate(s) if v !=0)
```

```python
import heapq, time

def astar(start, h):
    frontier = [(h(start), 0, start)]
    g = {start: 0}; parent = {start: (None, None)}; expanded = 0
    while frontier:
        f, gs, s = heapq.heappop(frontier)
        if gs > g[s]: continue
        expanded += 1
        if s == Goal:
            path = []
            while parent[s][0] is not None:
                s,a = parent[s]; path.append(a)
                return path[::-1], expanded, len(g)
        for  a, t in neighbours(s):
            if gs + 1 < g.get(t, float("inf")):
                g[t] = gs + 1
                parent[t] = (s, a)
                
                heapq.heappush(frontier, (g[t] + h(t), g[t], t))
```

```python
START = (7, 2, 4, 5, 0, 6, 8, 3, 1)
for name, h in [("UCS (h=0)", h_zero), ("A* misplaced",h_misplaced),
               ("A* Manhattan", h_mahattan)]:
    t0 = time.perf_counter(); path, exp, gen = astar(START, h)
    print(f"{name:14s} moves={len(path)} expanded={exp} generated = {gen}"
          f"time=:{time.perf_counter()-t0:.2f}s")
```
